  
## 1. 【第一部分】基礎函數定義與逐層前向傳播
本區塊定義激發函數、初始化參數，並將前向傳播的過程逐層計算並列印出對應的矩陣形狀與數值結果。

```python
import numpy as np

# 1. 定義所需的激發函數與損失函數
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

def cross_entropy_error(y_hat, y):
    delta = 1e-7
    return -np.sum(y * np.log(y_hat + delta))

# 2. 設定隨機種子以確保結果可重現
np.random.seed(42)

# 3. 自行產生輸入資料 x（具有 6 個特徵值的向量）
x = np.array([0.5, 1.2, -0.8, 2.1, 0.3, -1.5])

# 4. 初始化各層權重 W 與偏值 b
W1 = np.random.randn(6, 5) * 0.1
b1 = np.random.randn(5) * 0.1

W2 = np.random.randn(5, 4) * 0.1
b2 = np.random.randn(4) * 0.1

W3 = np.random.randn(4, 4) * 0.1
b3 = np.random.randn(4) * 0.1

W4 = np.random.randn(4, 3) * 0.1
b4 = np.random.randn(3) * 0.1

# 5. 設定真實標籤 y (此處設為 One-Hot 向量，假設正確答案為第 3 類)
y = np.array([0, 0, 1])

# 6. 列印出各層參數的形狀與個數
print("【各層參數形狀與個數註明】")
print(f"輸入向量 x -> 形狀: {x.shape}")
print(f"第一層 W1 -> 形狀: {W1.shape}, 參數個數: {W1.size} | b1 -> 形狀: {b1.shape}, 參數個數: {b1.size}")
print(f"第二層 W2 -> 形狀: {W2.shape}, 參數個數: {W2.size} | b2 -> 形狀: {b2.shape}, 參數個數: {b2.size}")
print(f"第三層 W3 -> 形狀: {W3.shape}, 參數個數: {W3.size} | b3 -> 形狀: {b3.shape}, 參數個數: {b3.size}")
print(f"輸出層 W4 -> 形狀: {W4.shape}, 參數個數: {W4.size} | b4 -> 形狀: {b4.shape}, 參數個數: {b4.size}")
print("=" * 60)
```

```python
# 執行前向傳播逐層計算
# 1. 第一隱藏層計算 (ReLU)
z1 = np.dot(x, W1) + b1
h1 = relu(z1)

# 2. 第二隱藏層計算 (ReLU)
z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

# 3. 第三隱藏層計算 (Sigmoid)
z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

# 4. 輸出層計算 (Softmax)
z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)

# 5. 列印前向傳播的矩陣計算結果與形狀
print("【前向傳播計算結果】\n")
print(f"第一隱藏層 z1: {z1}\n形狀: {z1.shape}")
print(f"第一隱藏層 h1 (ReLU): {h1}\n形狀: {h1.shape}\n")
print("-" * 50)
print(f"第二隱藏層 z2: {z2}\n形狀: {z2.shape}")
print(f"第二隱藏層 h2 (ReLU): {h2}\n形狀: {h2.shape}\n")
print("-" * 50)
print(f"第三隱藏層 z3: {z3}\n形狀: {z3.shape}")
print(f"第三隱藏層 h3 (Sigmoid): {h3}\n形狀: {h3.shape}\n")
print("-" * 50)
print(f"輸出層 z4: {z4}\n形狀: {z4.shape}")
print(f"輸出層 y_hat (Softmax 預測機率): {y_hat}\n形狀: {y_hat.shape}\n")
print("=" * 50)

# 6. 計算交叉熵誤差
loss = cross_entropy_error(y_hat, y)
print("【誤差評估結果】")
print(f"真實結果 y: {y}")
print(f"預測結果 y_hat: {y_hat}")
print(f"交叉熵誤差 (Cross Entropy Error): {loss:.6f}")
```

## 2. 【第二部分】封裝為 dnn 函數與 Gradio 互動介面
將上述網路設計封裝為單一函數 `dnn(x)`，並導入 `gradio` 套件建立互動式 Web 介面。使用者可自由輸入 6 個特徵值，介面會將輸出機率最高的類別轉化為「類別 1、類別 2 或 類別 3」呈現。

```python
# 安裝 Gradio 套件
!pip install gradio -q

import gradio as gr

# 依題目要求：將整個全連接神經網路設計為一個函數，名稱就稱為 dnn，且必須有一個參數 x
def dnn(x):
    z1 = np.dot(x, W1) + b1
    h1 = relu(z1)

    z2 = np.dot(h1, W2) + b2
    h2 = relu(z2)

    z3 = np.dot(h2, W3) + b3
    h3 = sigmoid(z3)

    z4 = np.dot(h3, W4) + b4
    y_hat_res = softmax(z4)
    return y_hat_res

# 驗證函數結果是否與第 1 項練習完全相同
y_hat_from_func = dnn(x)
print("【函數封裝驗證】")
print(f"由 dnn 函數算出的 y_hat：\n{y_hat_from_func}")
print(f"與前面逐層計算的 y_hat 是否完全相同：{np.allclose(y_hat, y_hat_from_func)}")
print("=" * 60)

# 定義給 Gradio 的呼叫函數
def gradio_predict(f1, f2, f3, f4, f5, f6):
    input_x = np.array([f1, f2, f3, f4, f5, f6])
    y_hat_val = dnn(input_x)
    predicted_class = int(np.argmax(y_hat_val) + 1)
    return f"類別 {predicted_class}"

# 設計 Gradio 介面
demo = gr.Interface(
    fn=gradio_predict,
    inputs=[
        gr.Number(value=0.5, label="特徵 1 (x1)"),
        gr.Number(value=1.2, label="特徵 2 (x2)"),
        gr.Number(value=-0.8, label="特徵 3 (x3)"),
        gr.Number(value=2.1, label="特徵 4 (x4)"),
        gr.Number(value=0.3, label="特徵 5 (x5)"),
        gr.Number(value=-1.5, label="特徵 6 (x6)"),
    ],
    outputs=gr.Text(label="預測機率最高的類別"),
    title="DNN 前向傳播與 Gradio 互動介面",
    description="請輸入 6 個特徵值，模型將自動計算前向傳播，並返回機率最高的類別（1, 2, 或 3）。"
)

# 啟動互動網頁
demo.launch(share=True)
```
